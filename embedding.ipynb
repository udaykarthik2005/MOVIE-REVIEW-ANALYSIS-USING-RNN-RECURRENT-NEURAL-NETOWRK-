```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[
'the glass of milk',
'the glass of juice',
'the cup of tea',
'I am a good boy',
'I am a good developer',
'understand the meaning of words',
'your videos are good',]
```

```python
## define vocabulary size
voc_size= 10000
```

```python
## one hot representation for every word
one_hot_repr = [one_hot(words,voc_size)for words in sent]
one_hot_repr
```

```python
## word embedding representation
from tensorflow.keras.layers import Embedding 
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential 
```

```python
import numpy as np
```

```python
sent_length = 8
Embedding_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(Embedding_docs)
```

```python
## feature representation
dim = 10
```

```python
## embedding model
model = Sequential()
model.add(Embedding(input_dim=voc_size,
                    output_dim=dim,
                    input_shape=(sent_length,)))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(Embedding_docs)
##every word gets represented by 10 dimensions
```
